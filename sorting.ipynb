```python
import numpy as np
```

```python
test_arr = [3,8,7,1,2,3,2,8,10,14,2,1,6,123,54,23,12,55,32,1,2,13]
```

**Bubble-Sort**

```python
def bubble_sort(arr):
    l = len(arr)
    for i in range(l-1):
        finished = True
        for j in range(l-1-i):
            if arr[j]>arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                finished = False
        if finished == True:
            break
    return arr
```

```python
bubble_sort(test_arr)
```

**Merge-Sort**

```python
def merge(arr1,arr2):
    merged = []

    l1 = len(arr1)
    l2 = len(arr2)
    tl = l1+l2
    
    c1 = 0
    c2 = 0
    while len(merged)<tl:
        if arr1[c1]>arr2[c2]:
            merged.append(arr2[c2])
            c2+=1
        else:
            merged.append(arr1[c1])
            c1+=1

        if c2==l2:
            merged+=arr1[c1:]
        elif c1==l1:
            merged+=arr2[c2:]

    return merged

def merge_sort(arr):
    if len(arr)<=2:
        return arr
    
    mid = len(arr)//2

    return merge(merge_sort(arr[:mid]),merge_sort(arr[mid:]))
```

```python
merge_sort(test_arr)
```

**Quick-Sort**

```python
def quick_sort(arr):
    if len(arr)<=2:
        return arr

    mid_id = np.random.randint(0,len(arr), (1,))
    mid_element = arr[mid_id[0]]

    low = []
    mid = []
    high = []

    for i in arr:
        if i<mid_element:
            low.append(i)
        if i==mid_element:
            mid.append(i)
        if i>mid_element:
            high.append(i)

    return quick_sort(low)+mid+quick_sort(high)
```

```python
quick_sort(test_arr)
```
